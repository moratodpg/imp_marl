## Heuristic policy

```python
def heuristic(timestep, inspections, pf_comp, inspection_interval, n_components):
    actions = np.zeros(len(pf_comp), dtype=int)

    if timestep % inspection_interval == 0 and timestep > 0:
        sorted_indices = np.argsort(pf_comp)[::-1]
        sorted_indices = sorted_indices[:n_components]
        actions[sorted_indices] = 2  # Inspect

    # assign action 4 for all indices with inspections equal to 1
    actions[inspections == 1] = 4  # Repair

    print(f"timestep: {timestep}, actions: {actions}, pf_comp: {pf_comp}, inspections: {inspections}")

    return actions.tolist()
```

```python
observations = env.reset()
print(observations['agent_0'][-5:])
done = False
rewards = 0
inspections = np.ones(len(env.agent_list), dtype=int)*61
while not done:
    print(env.time_step)
    print(observations['agent_0'][-5:])
    pf_comps = env.damage_proba[0, :].reshape((env.lev, env.stress_conditions, env.crack_conditions)).sum(axis=1)[:, -1]
    actions = heuristic(env.time_step, inspections, pf_comps[:2], 1, 2)
    # convert actions to dict
    actions = {agent: actions[i] for i, agent in enumerate(env.agent_list)}
    observations, reward, done, info = env.step(actions)
    inspections = info["inspections"][0, :2]
    print(info)
    print("rewards:", reward)
    rewards += reward['agent_0']
```

## Heur monitoring

```python
def heuristic_monitoring(timestep, inspections, pf_comp, inspection_interval, n_components):
    actions = np.zeros(len(pf_comp), dtype=int)

    if timestep % inspection_interval == 0 and timestep > 0:
        sorted_indices = np.argsort(pf_comp)[::-1]
        sorted_indices = sorted_indices[:n_components]
        actions[sorted_indices] = 1  # Install sensor

    # assign action 4 for all indices with inspections more than 45 and less than 61
    actions[(inspections > 15) & (inspections < 61)] = 4  # Repair

    # print(f"timestep: {timestep}, actions: {actions}, pf_comp: {pf_comp}, inspections: {inspections}")

    return actions.tolist()
```

```python
observations = env.reset()
print(observations['agent_0'][-5:])
done = False
rewards = 0
inspections = np.ones(len(env.agent_list), dtype=int)*61
while not done:
    print(env.time_step)
    print(observations['agent_0'][-5:])
    pf_comps = env.damage_proba[0, :].reshape((env.lev, env.stress_conditions, env.crack_conditions)).sum(axis=1)[:, -1]
    actions = heuristic_monitoring(env.time_step, inspections, pf_comps[:2], 1, 2)
    # convert actions to dict
    actions = {agent: actions[i] for i, agent in enumerate(env.agent_list)}
    observations, reward, done, info = env.step(actions)
    inspections = info["inspections"][0, :2]
    print(info)
    print("rewards:", reward)
    rewards += reward['agent_0']
```

```python
start_time = time.time()

rewards_eps = []
eps = 1000

for ep in range(eps):
    observations = env.reset()
    done = False
    rewards = 0
    inspections = np.ones(len(env.agent_list), dtype=int)*61
    while not done:
        pf_comps = env.damage_proba[0, :].reshape((env.lev, env.stress_conditions, env.crack_conditions)).sum(axis=1)[:, -1]
        actions = heuristic_monitoring(env.time_step, inspections, pf_comps[:2], 1, 2)
        # convert actions to dict
        actions = {agent: actions[i] for i, agent in enumerate(env.agent_list)}
        observations, reward, done, info = env.step(actions)
        inspections = info["inspections"][0, :2]
        rewards += reward['agent_0']
    rewards_eps.append(rewards)

end_time = time.time()
print(f"Total time: {end_time - start_time} seconds")

print(f"Average rewards per episode: {np.mean(rewards_eps)}")
```

```python
# generate 100,000 gaussian with different means and std
n_samples = 100000
means = np.random.uniform(-10, 10, n_samples)
stds = np.random.uniform(1, 5, n_samples)
```

```python
samples_outer_loop = 1000
# repeat over samples_outer_loop
pf_all = []
for _ in range(samples_outer_loop):
    # sample each gaussian and count if it is negative
    negative_samples = 0
    for i in range(n_samples):
        sample = np.random.normal(loc=means[i], scale=stds[i])
        if sample < 0:
            negative_samples += 1
    pf_ = negative_samples / n_samples
    pf_all.append(pf_)
```

```python
# count mean and std from pf_all
pf_mean = np.mean(pf_all)
pf_std = np.std(pf_all)

print(f"Mean of pf: {pf_mean}")
print(f"Std of pf: {pf_std}")
```

```python
import math
def norm_cdf(mu=0.0, sigma=1.0):
    z = (- mu) / sigma
    return 0.5 * (1.0 + math.erf(z / math.sqrt(2.0)))
```

```python
# iterate over all means and stds (not samples) and compute norm cdf
pf_norm_cdf = [norm_cdf(mu=mean_, sigma=std_) for mean_, std_ in zip(means, stds)]
pf_norm_ = np.array(pf_norm_cdf)
```

```python
# compute mean and std of pf_norm_cdf
pf_mean_cf = pf_norm_.mean()
pf_std_cf  = np.sqrt((pf_norm_ * (1 - pf_norm_)).sum()) / len(pf_norm_)

print(f"Mean of pf: {pf_mean_cf}")
print(f"Std of pf: {pf_std_cf}")
```
